## Project Problem

## Workflow Explanation

1. **Data Collection:** University information is collected from reliable sources such as official documents, webpages, FAQs, and other institutional resources.

2. **Data Preprocessing:** The collected text is cleaned and prepared by removing unnecessary information and formatting it into usable text.

3. **Text Chunking and Embedding:** Large documents are divided into smaller sections. Each section is converted into a numerical vector representation called an embedding.

4. **Vector Store:** The embeddings are stored in a vector database or vector index so that relevant information can be retrieved efficiently.

5. **Question Processing:** When a user submits a question, the question is processed and converted into an embedding using the same or a compatible embedding model.

6. **Semantic Retrieval:** The question embedding is compared with the stored document embeddings. The system retrieves the sections that are most semantically similar to the question.

7. **Answer Generation:** The retrieved information is provided to an AI language model, which generates an answer based on the relevant information.

8. **Response:** The final answer is presented to the user in a natural-language format.

This workflow combines machine learning, Natural Language Processing, semantic search, and retrieval-augmented generation (RAG) to improve the relevance and reliability of chatbot responses.

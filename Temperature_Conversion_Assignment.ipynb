```python
# Python and Computing Environment Set-up - Temperature Conversion and Threshold Assignment

This notebook demonstrates temperature conversion from Celsius to Fahrenheit,
threshold checking, loops, and the use of a Python function.
```

```python
# Temperature Conversion Assignment
# This program converts temperatures from Celsius to Fahrenheit,
# checks whether a temperature is above or below a specified threshold,
# uses a loop to convert several temperatures, and uses a function.


# Function to convert Celsius to Fahrenheit
def celsius_to_fahrenheit(celsius):
    """Convert a temperature from Celsius to Fahrenheit."""
    fahrenheit = (celsius * 9 / 5) + 32
    return fahrenheit


# Ask the user to enter a temperature
celsius = float(input("Enter a temperature in °C: "))

# Convert the temperature using the function
fahrenheit = celsius_to_fahrenheit(celsius)

# Display the converted temperature
print(f"{celsius}°C = {fahrenheit:.2f}°F")


# Specify the threshold temperature
threshold = 30

# Check whether the temperature is above or below the threshold
if celsius > threshold:
    print(f"The temperature is above the threshold of {threshold}°C.")
elif celsius < threshold:
    print(f"The temperature is below the threshold of {threshold}°C.")
else:
    print(f"The temperature is exactly at the threshold of {threshold}°C.")


# Use a loop to convert several temperatures
print("\nConverting several temperatures:")

temperatures = [0, 10, 20, 25, 30, 40]

for temp in temperatures:
    converted = celsius_to_fahrenheit(temp)
    print(f"{temp}°C = {converted:.2f}°F")
```
